### Ejercicio 8:
¿Cuántos tickets y qué importe total se registraron en cada mes? ¿En qué mes se alcanzó el mayor importe?

```python
import pandas as pd
datosVentas = pd.read_csv("ventas.csv")
```

```python
datosVentas["importe_linea"] = (
    datosVentas["cantidad"] * datosVentas["precio_unitario"]
)

datosVentas["mes"] = (
    pd.to_datetime(datosVentas["fecha"]).dt.to_period("M").astype(str)
)

ticketsPorMes = datosVentas.groupby("mes")["id_ticket"].nunique()
importePorMes = datosVentas.groupby("mes")["importe_linea"].sum()

tabla8 = pd.DataFrame({
    "Mes": ticketsPorMes.index,
    "Cantidad de tickets": ticketsPorMes.values,
    "Importe total": importePorMes.values
})

tabla8["Importe total"] = tabla8["Importe total"].map("${:.2f}".format)
display(tabla8)

print("Mes con mayor importe:", importePorMes.idxmax())
print("Importe de ese mes: ${:.2f}".format(importePorMes.max()))
```
